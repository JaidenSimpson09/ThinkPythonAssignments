```python
#Import statements

#FUNCTION DEFINITIONS
#calculate_rectangle
def calculate_rectangle(length, width):
  #Accept length and width as paramiters
  #Calculate area by multiplying length * width
  area = length * width
  #Calculate paramiter by multiplying 2 * (length + width)
  perimeter = 2 * (length + width)
  #Put both values into a tuple (area, perimeter)
  #Return the tuple
  return (area, perimeter)

#PROGRAM LOGIC
#Turn imputs into floats
#Ask the user for length
length = float(input("Enter Length: "))
#Ask the user for width
width = float(input("Enter Width: "))

#Call calculate_rectangle function
#Unpack the returned tuple into my_area and my_perimeter
my_area, my_perimeter = calculate_rectangle(length, width)

#Print results
print("\n RESULTS")
print("the area is", my_area)
print("The parimeter is", my_perimeter)
```

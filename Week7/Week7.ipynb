```python
# FUNCTION DEFINITIONS

# 1. Define verify_pin():
def verify_pin():
#    - Create a loop that asks for a PIN until it equals "1234"
  while True:
#    - Return True
    pin = input ("Enter Pin: ")

    if pin =="1234":
      print ("PIN Verfied.")
      return True

    else:
      print ("Incorrect Pin. Try Again")

  verify_pin()

# 2. Define withdraw(balance, amount):
def withdraw (balance, amount):
#    - Check if enough money exists
  if amount <= balance:
    balance -= amount
    print ("Withdrawl Successful")
  else:
    print ("Insufficient Funds")


#    - Return the new (or old) balance
  return balance

# PROGRAM LOGIC

# 1. Set current_balance to 500
current_balance = 500

# 2. Call verify_pin()
verify_pin()

# 3. Get withdrawal amount from user input
amount = float(input("Enter Withdrawl Amount: "))

# 4. Update current_balance by calling withdraw()
current_balance = withdraw(current_balance, amount)

# 5. Print the final balance
print ("Current Balance: $" + str(round(current_balance, 2)))
```

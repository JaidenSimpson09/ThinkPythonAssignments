```python
#IMPORT STATMENTS

#FUNCTION DEFINITIONS
  #add_players
    #get names from user
    #stop when user types "quit"
    #add names to list

def add_players(roster):
  while True:
    name = input("Enter Player Name(or 'quit' to stop): ")

    if name.lower()== "quit":
      break

    roster.append(name)



  #is_on_roster
    #check if name is on the list
    #return true or false

def is_on_roster(roster, search_name):
  for player in roster:
    if player.lower() == search_name.lower():
      return True
  return False

#PROGRAM LOGIC
  #make empty list
roster = []
  #call add players
add_players(roster)
  #ask for name to search
search_name = input("Who would you like to search for? ")
  #call is_on_roster
found = is_on_roster(roster, search_name)

  #print result
if found == True:
  print("Result: Player found!")
else:
  print("Result: Player not found.")
```

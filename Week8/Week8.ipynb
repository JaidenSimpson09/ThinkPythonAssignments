```python
# Import statements


# Functions Definitions
  # Make a funtion called is_palindrome that takes one word as a pramater.
  # inside the function
    # change the word to lowercase so uppercase and lowercase letters dont matter.
    # reverse the word using slicing
    # check if the origional word and the reversed word are the same.
    # if they are not the same then return true
    # if they are the same, then return false
def is_palindrome (word):
    word = word.lower()
    reversed_word = word [::-1]

    if word == reversed_word:
      return True
    else:
      return False

# Program logic
  #Ask the user to type the word
  #save the word in a varible

user_word = input("Enter a word: ")

  #Call is_palindrome function and send the user's word to it
  #save the answer (true or false)

result = is_palindrome(user_word)

  #if the answer is true, print that the word is a palindrome
  #if the answer is false, print that the word not a palindrome

if result == True:
  print (user_word + " is a palindrome")
else:
  print (user_word + " is not a palindrome")

```

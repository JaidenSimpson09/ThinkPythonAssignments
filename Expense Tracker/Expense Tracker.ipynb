```python
#MAIN LOOP

# Data structure:
  # Use a list to store all expenses
expenses = []

# Repeat until user chooses exit
while True:

# Display menu options (Add expense, Get totals, Exit)
  print("\n Expense Tracker")
  print("1. Add Expense")
  print("2. Get Totals")
  print("3. Exit")

# Ask user for selection
  choice = input("Selection: ")

# If the user selects add expense:
  if choice == "1":

  # Prompt for category, vendor, amount, date
    category = input("Enter Category: ")
    vendor = input("Enter Vendor: ")

    while True:
      amount_input = input("Enter Amount: ")
      try:
        amount = float(amount_input)
        break
      except:
         print("Invalid amount. Please enter a number.")

    date = input ("Enter Date: ")

  # Store as a dictionary
    expense = {
        "category": category,
        "vendor": vendor,
        "amount": amount,
        "date": date}
  # Add dictionary to a list
    expenses.append(expense)

    print(expenses)
# If the user selects Get totals:
  elif choice == "2":
    # Show a sub menu (All, Category, Vendor)
    print("\n Totaling Options")
    print("A: All Expenses")
    print("C: Category")
    print("V: Vendor")

    # Perform filtering and sum amounts
    option = input("Choice: ")

    if option.upper() == "A":
        total = 0
        for expense in expenses:
          total += expense["amount"]
        print(f"Total of all expenses: ${total:.2f}")

    elif option.upper() == "C":
        category_input = input ("Enter Category: ")
        total = 0
        for expense in expenses:
            if expense ["category"].lower() == category_input.lower():
                total += expense["amount"]
        print(f"The total for {category_input} is: ${total:.2f}")

    elif option.upper() == "V":
        vendor_input = input("Enter vendor: ")
        total = 0
        for expense in expenses:
            if expense["vendor"].lower() == vendor_input.lower():
                total += expense ["amount"]
        print(f"The total for {vendor_input} is: ${total:.2f}")




  elif choice == "3":
    break



# Each expense is a dictionary with:
  # (category, vendor, amount, and date)

# Filtering logic:
  # loop through list of expenses
  # Compare category or vendor to user input
  # Add matching amounts to total
```

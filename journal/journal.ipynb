```python
#IMPORT STATEMENTS

#FUNCTION DEFINITIONS

#write_entry
def write_entry(filename, entry):

  # Accept filename and entry as parameters
  # Open file in append mode ('a')
  file = open(filename, 'a')

  # Write entry to file
  # Make sure each entry goes on a new line using \n
  file.write (entry + "\n")

  # Close the file
  file.close()

#read_journal
def read_journal(filename):

  # Accept filename as a parameter
  try:

  # Open file in read mode ('r')
    file = open(filename, "r")

  # Read all contents
    content = file.read()

  # Print contents to screen
    print("Your Journal")
    print()
    print(content)
    file.close()

  except FileNotFoundError:
    print("No journal entries found yet.")
#PROGRAM LOGIC
filename = "journal.txt"

while True:

# Ask user if they want to write, read or quit
  choice = input("(W)rite, (R)ead, or (Q)uit?")

# If W, get entry and call write_entry
  if choice.lower() == "w":
    entry = input("Enter your entry")
    write_entry(filename, entry)
    print("entry saved.")

# If R, call read_journal
  elif choice.lower() == "r":
    read_journal(filename)

# Repeat using a loop untill the user types "quit"
  elif choice.lower() == "q":
    break
```

```python
#Import Statments

#Function Definitions
  # validate_code, this makes sure the user only puts B, S or L
def validate_code():

  code = input("Enter Classification Code (B, S, L): ")
  code = code.upper()

  while code != "B" and code != "S" and code != "L":

    print("Invalid code. Please enter B, S or L: ")
    code = input("Enter Classification Code (B, S, L): ")
    code = code.upper()

  return code

  # calculate_total_miles, Subtract the start odometer from the end odometer
def calculate_total_miles (start, end):

  miles = end - start

  return miles
  # calculate_charged_miles, This will find out how many miles will be charged and the first 100 miles are free
def calculate_charged_miles (total_miles):

  if total_miles > 100:

    charged = total_miles - 100

  else:

    charged = 0


  return charged


  # get_classification_name, Turn B, S or L into the full car type name
def get_classification_name (code):
    if code == "B":
      name = "Budget"
    elif code == "S":
      name = "Standard"

    else:
      name = "Luxury"

    return name

  # calculate_total_bill, this will calculate the total bill using the daily price and milage cost
def calculate_total_bill (code, days, charged_miles):

  if code == "B":
    daily_rate = 40
    mile_rate = 0.25

  elif code == "S":
    daily_rate = 60
    mile_rate = 0.25

  else:
    daily_rate = 100
    mile_rate = 0.25

  total = (days * daily_rate) + (charged_miles * mile_rate)

  return total




#Program Logic
  #ask the user for classification code and make sure it is B, S or L
code = validate_code()

  # Ask the user for start and end odometer
start_odometer = int(input("Start Odometer:"))

end_odometer = int(input("End Odometer: "))

  # ask the user for total miles driven
total_miles = calculate_total_miles(start_odometer, end_odometer)

  # Ask user for number of days rented
days_rented = int(input("Days Rented: "))

  # Calculate the miles charged and total bill
charged_miles = calculate_charged_miles(total_miles)
total_bill = calculate_total_bill(code, days_rented, charged_miles)

  # Print rental summary
classification_name = get_classification_name(code)
print()
print("RENTAL SUMMARY")
print("Classification:",classification_name)
print("Total Miles:",total_miles)
print("Charged Miles:",charged_miles)
print("Total Bill: $",total_bill)
```
